```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Type of plots

#### Line plots

```python
x = [1,2,3,4,5]
y = [10,20,25,30,40]
```

```python
plt.plot(x,y);
plt.title("Simple Line Plot")
plt.xlabel('x-axis label')
plt.ylabel('y-axis label')
plt.grid(True)
```

#### Bar Chart

```python
months = ['jan','Feb','Mar','Apr']
revenue = [100,200,300,500]
```

```python
plt.figure(figsize=(10,16))
plt.bar(months,revenue)
plt.title("Monthly Revenue Bar chart")
plt.xlabel('Months')
plt.ylabel('revenue')
```

#### Histogram

```python
marks50 = np.random.randint(0,100,50)
marks50
```

```python
plt.hist(marks50)
plt.xlabel("Marks")
plt.ylabel("No. of Student")
```

```python
plt.hist(marks50,bins=2)
```

#### Scatteer plot

```python
heights = [65,70,68,72,66]
weights = [150,160,155,180,145]
```

```python
plt.scatter(heights,weights)
plt.title('Height v/s Weights Scatter Plot')
plt.xlabel('Height (inches)')
plt.ylabel('Height (pounds)')
```

#### SubPlots and Legends

```python
days= [1,2,3,4,5]
original_price = [10,20,25,30,40]
predicted_price = [8,21,25,31,42]
```

```python
plt.plot(days,original_price,'bo-',label='Original Price')
plt.plot(days,predicted_price,'ro-',label='Predicted Price')
plt.legend()
plt.title('Plot with Legend')
plt.xlabel('weekdays')
plt.ylabel('Price')
```

#### SubPlots

```python
# first subplot - original price
plt.subplot(3,1,1)
plt.plot(days,original_price,'b-o',label='original price')
plt.title('Original Price Over time')
plt.xlabel('weekdays')
plt.ylabel('Price')
plt.legend()

# second subplot - original price
plt.subplot(3,1,2)
plt.plot(days,predicted_price,'ro-',label='Predicted Price')
plt.title('Predicted Price Over time')
plt.xlabel('weekdays')
plt.ylabel('Price')
plt.legend()

# Third subplot - original price
price_difference = [0-p for o,p in zip(original_price,predicted_price)]
plt.subplot(3,1,3)
plt.bar(days,price_difference,color='purple',label='Predicted Price')
plt.title('Difference between Original & Predicted Price')
plt.xlabel('Days')
plt.ylabel('Price Difference')
plt.legend()

plt.tight_layout()
```

#### Saving the figure

```python
plt.plot(days,original_price,'bo-',label='Original Price')
plt.plot(days,predicted_price,'ro-',label='Predicted Price')
plt.legend()
plt.title('Plot with Legend')
plt.xlabel('weekdays')
plt.ylabel('Price')
plt.savefig('example_prediction.png',dpi=300) #save the image
```
